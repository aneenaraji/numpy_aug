```python
# create a 3•4 matrxix with float values
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z)

#floor-previous, tazhthe integer round aaku
print(np.floor(z))

#ceil-next, upper integer round aaku
print(np.ceil(z))

print(np.round(z)) #nearest even if .5 comes it goes to nearest even
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max())

print(z.min())

print(z.sum())

#first row max element
print(z.max(axis=1)) #to get row in numpy, opp to pandas

#lowest element in row- numpy always give row , pandas column
print(z.max(axis=0)) #to get column

#to get square root
print(np.sqrt(z))

print(np.sin(z))

print(np.cos(z))
```
